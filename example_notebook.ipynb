# Análisis de Redes Cliente-Servicio: Comparación de Períodos

Este notebook demuestra el uso del módulo de análisis de redes cliente-servicio para comparar redes en diferentes períodos temporales (pre vs. post evento).

```python
# Importar módulos necesarios
from network_analysis import (
    ClientServiceNetwork,
    NetworkMetrics,
    NetworkComparison,
    NetworkVisualization,
    comprehensive_period_comparison
)
import matplotlib.pyplot as plt
import pandas as pd

# Configurar visualización
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 8)
```

## 1. Crear Red Pre-Evento

Primero, creamos una red que representa el estado antes del evento.

```python
# Crear red pre-evento
network_pre = ClientServiceNetwork(weighted=True)

# Agregar servicios con metadatos
services = [
    ('S1', {'name': 'Web Server', 'capacity': 100, 'type': 'web'}),
    ('S2', {'name': 'Database', 'capacity': 80, 'type': 'database'}),
    ('S3', {'name': 'API Gateway', 'capacity': 90, 'type': 'api'}),
    ('S4', {'name': 'Cache Server', 'capacity': 70, 'type': 'cache'}),
    ('S5', {'name': 'Auth Service', 'capacity': 85, 'type': 'auth'}),
]

for service_id, metadata in services:
    network_pre.add_service(service_id, metadata)

# Agregar clientes
clients = ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8']
for client_id in clients:
    network_pre.add_client(client_id)

# Agregar conexiones
connections = [
    ('C1', 'S1', 5.0),
    ('C1', 'S2', 3.0),
    ('C2', 'S1', 4.0),
    ('C2', 'S3', 6.0),
    ('C3', 'S2', 7.0),
    ('C3', 'S4', 2.0),
    ('C4', 'S3', 5.0),
    ('C4', 'S5', 4.0),
    ('C5', 'S1', 3.0),
    ('C5', 'S4', 5.0),
    ('C6', 'S2', 4.0),
    ('C6', 'S5', 6.0),
    ('C7', 'S3', 3.0),
    ('C7', 'S4', 4.0),
    ('C8', 'S5', 5.0),
]

for client, service, weight in connections:
    network_pre.add_connection(client, service, weight)

print(f"Red pre-evento creada:")
print(f"  - Nodos: {network_pre.number_of_nodes()}")
print(f"  - Enlaces: {network_pre.number_of_edges()}")
```

## 2. Visualizar Red Pre-Evento

```python
NetworkVisualization.plot_network(network_pre, "Red Pre-Evento")
plt.show()
```

## 3. Crear Red Post-Evento

Ahora creamos una red que representa el estado después del evento, con cambios estructurales.

```python
# Crear red post-evento
network_post = ClientServiceNetwork(weighted=True)

# Agregar servicios (incluyendo uno nuevo)
services_post = [
    ('S1', {'name': 'Web Server', 'capacity': 120, 'type': 'web'}),
    ('S2', {'name': 'Database', 'capacity': 100, 'type': 'database'}),
    ('S3', {'name': 'API Gateway', 'capacity': 110, 'type': 'api'}),
    ('S4', {'name': 'Cache Server', 'capacity': 80, 'type': 'cache'}),
    ('S5', {'name': 'Auth Service', 'capacity': 95, 'type': 'auth'}),
    ('S6', {'name': 'Load Balancer', 'capacity': 100, 'type': 'load_balancer'}),
]

for service_id, metadata in services_post:
    network_post.add_service(service_id, metadata)

# Agregar clientes (incluyendo nuevos)
clients_post = ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']
for client_id in clients_post:
    network_post.add_client(client_id)

# Agregar conexiones con patrones actualizados
connections_post = [
    ('C1', 'S1', 6.0),
    ('C1', 'S2', 4.0),
    ('C1', 'S6', 5.0),
    ('C2', 'S1', 5.0),
    ('C2', 'S3', 7.0),
    ('C3', 'S2', 8.0),
    ('C3', 'S4', 3.0),
    ('C3', 'S6', 4.0),
    ('C4', 'S3', 6.0),
    ('C4', 'S5', 5.0),
    ('C5', 'S1', 4.0),
    ('C5', 'S4', 6.0),
    ('C5', 'S6', 5.0),
    ('C6', 'S2', 5.0),
    ('C6', 'S5', 7.0),
    ('C7', 'S3', 4.0),
    ('C7', 'S4', 5.0),
    ('C8', 'S5', 6.0),
    ('C8', 'S6', 5.0),
    ('C9', 'S1', 4.0),
    ('C9', 'S6', 5.0),
    ('C10', 'S2', 5.0),
    ('C10', 'S3', 6.0),
]

for client, service, weight in connections_post:
    network_post.add_connection(client, service, weight)

print(f"Red post-evento creada:")
print(f"  - Nodos: {network_post.number_of_nodes()}")
print(f"  - Enlaces: {network_post.number_of_edges()}")
```

## 4. Visualizar Red Post-Evento

```python
NetworkVisualization.plot_network(network_post, "Red Post-Evento")
plt.show()
```

## 5. Análisis Comparativo Completo

Realizamos un análisis comparativo completo entre las dos redes.

```python
# Realizar comparación completa
results = comprehensive_period_comparison(network_pre, network_post, "notebook_output")
```

## 6. Análisis de Centralidad

Comparamos las métricas de centralidad entre ambos períodos.

```python
# Visualizar comparación de centralidad
centrality_df = results['centrality']
print("\nTop 10 nodos por cambio en centralidad de grado:")
print(centrality_df.nlargest(10, 'degree_change')[['node', 'degree_pre', 'degree_post', 'degree_change', 'degree_change_pct']])

# Graficar comparación
NetworkVisualization.plot_centrality_comparison(centrality_df, 'degree', top_n=10)
plt.show()
```

## 7. Análisis de Conectividad

Examinamos los cambios en las métricas de conectividad.

```python
connectivity_comparison = results['connectivity']

print("\nMétricas de Conectividad:")
print("\nPre-evento:")
for key, value in connectivity_comparison['pre'].items():
    print(f"  {key}: {value}")

print("\nPost-evento:")
for key, value in connectivity_comparison['post'].items():
    print(f"  {key}: {value}")

print("\nCambios:")
for key, value in connectivity_comparison['changes'].items():
    if isinstance(value, float):
        print(f"  {key}: {value:+.4f}")
    else:
        print(f"  {key}: {value:+d}")
```

## 8. Análisis de Comunidades

Comparamos la estructura comunitaria entre ambos períodos.

```python
community_comparison = results['communities']

print("\nAnálisis de Comunidades:")
print(f"Comunidades pre-evento: {community_comparison['num_communities_pre']}")
print(f"Comunidades post-evento: {community_comparison['num_communities_post']}")
print(f"Cambio: {community_comparison['community_change']:+d}")
print(f"Nodos con cambio de comunidad: {community_comparison['nodes_with_different_community']}")

# Visualizar comunidades
communities_pre = community_comparison['communities_pre']
communities_post = community_comparison['communities_post']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))
plt.sca(ax1)
NetworkVisualization.plot_community_structure(network_pre, communities_pre, "Comunidades Pre-Evento")
plt.sca(ax2)
NetworkVisualization.plot_community_structure(network_post, communities_post, "Comunidades Post-Evento")
plt.show()
```

## 9. Transformaciones Estructurales

Identificamos los cambios estructurales clave.

```python
structural_changes = results['structural_changes']

print("\nTransformaciones Estructurales:")
print(f"\nCrecimiento de la red:")
print(f"  Nodos agregados: {structural_changes['nodes_added']}")
print(f"  Nodos removidos: {structural_changes['nodes_removed']}")
print(f"  Enlaces agregados: {structural_changes['edges_added']}")
print(f"  Enlaces removidos: {structural_changes['edges_removed']}")
print(f"  Cambio en densidad: {structural_changes['density_change']:.4f}")

if 'top_nodes_increasing_centrality' in structural_changes:
    print("\nTop nodos con aumento de centralidad:")
    for item in structural_changes['top_nodes_increasing_centrality']:
        print(f"  {item['node']}: +{item['degree_change']:.4f}")

if 'top_nodes_decreasing_centrality' in structural_changes:
    print("\nTop nodos con disminución de centralidad:")
    for item in structural_changes['top_nodes_decreasing_centrality']:
        print(f"  {item['node']}: {item['degree_change']:.4f}")
```
